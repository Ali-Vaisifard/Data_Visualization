```python
# importing visualization 
import matplotlib.pyplot as plt
import seaborn as sns
# Get Titanic Dataset
titanic = sns.load_dataset('titanic')
```

```python
print(titanic.head())
```

```python
# line plot
sns.lineplot(x='age', y='survived', data=titanic)
plt.title('Age vs Survival')
plt.xlabel('Age')
plt.ylabel('Survival')
plt.show()
```

```python
# histogram for survoval and age bins
sns.histplot(data=titanic, x='age', hue='survived', bins=10, kde= True)
plt.title('Age Distribution by Survival')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

```
